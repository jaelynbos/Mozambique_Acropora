```r
# Compute distance matrices
dna_dist1<-dist.dna(mrna1, model = "K80")
dna_dist2<-dist.dna(mrna2, model = "K80")
dna_dist3<-dist.dna(mrna3, model = "K80")
dna_dist4<-dist.dna(mrna4, model = "K80")
dna_dist5<-dist.dna(mrna5, model = "K80")
dna_dist6<-dist.dna(mrna6, model = "K80")
dna_dist7<-dist.dna(mrna7, model = "K80")
dna_dist8<-dist.dna(mrna8, model = "K80")
dna_dist9<-dist.dna(mrna9, model = "K80")
dna_dist10<-dist.dna(mrna10, model = "K80")
dna_dist11<-dist.dna(mrna11, model = "K80")
dna_dist12<-dist.dna(mrna12, model = "K80")
dna_dist13<-dist.dna(mrna13, model = "K80")
dna_dist14<-dist.dna(mrna14, model = "K80")
```

```r
# Build balanced minimum evolution trees
tree1<-fastme.bal(dna_dist1)
p1 <- ggtree(tree1) +
  geom_tiplab(size = 8) +
  theme_tree2() +
  theme(axis.title.x = element_blank(),
        axis.text.x = element_blank(),
        axis.ticks.x = element_blank(),
        axis.line.x = element_blank()) +
  xlim(0, max(node.depth.edgelength(tree1)) * 1.8)
```
