```r
#Load packages
library(tidyverse)
library(vcfR)
library(adegenet)
library(hierfstat)
```

```r
#Load in SNP data in VCF format
hes1_snps <- read.vcfR("/scratch/jbos/Moz_aligned_mil/angsd_moz_bcf/hes1.vcf.gz")
```

```r
#Convert to genind format
genind_snps <- vcfR2genind(hes1_snps)
```

```r
bamnames<-function(bam){
    a<-strsplit(bam,'files/')[[1]][2]
    b<-strsplit(a,'.s')[[1]][1]
    return(b)
    }
```

```r
rownames(genind_snps@tab)<-sapply(rownames(genind_snps@tab),FUN=bamnames)
```

```r
sppB_bams<-read.table('/home/jbos/Moz_reads/bam_names_grp1.txt')
sppB<-sapply(sppB_bams$V1,FUN=bamnames)
```

```r
sppB_snps<-genind_snps@tab[rownames(genind_snps@tab) %in% sppB,]
sppB_snps<-sppB_snps[!is.na(sppB_snps[,1]),]
sppB_snps<-as.data.frame(sppB_snps)
```

```r
sppA_snps<-genind_snps@tab[!(rownames(genind_snps@tab) %in% sppB),]
sppA_snps<-sppA_snps[!is.na(sppA_snps[,1]),]
sppA_snps<-as.data.frame(sppA_snps)
```
