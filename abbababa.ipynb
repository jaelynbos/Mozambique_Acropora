```r
ab_AmSam<-read.table('/scratch/jbos/abbababa/results_AmSam.Angsd.abbababa2',header=TRUE)
ab_AmSam_chroms<-ab_AmSam[substr(ab_AmSam$CHR,1,2)=='NC',]
```

```r
HD_HE_MozB<-ab_AmSam_chroms[seq(1,nrow(ab_AmSam_chroms),3),]
D1<-HD_HE_MozB$Numer/HD_HE_MozB$Denom
```

```r
HD_MozB_HE<-ab_AmSam_chroms[seq(2,nrow(ab_AmSam_chroms),3),]
D2<-HD_MozB_HE$Numer/HD_MozB_HE$Denom
```

```r
HE_MozB_HD<-ab_AmSam_chroms[seq(3,nrow(ab_AmSam_chroms),3),]
D3<-HE_MozB_HD$Numer/HE_MozB_HD$Denom
```

```r
HD_MozB_HE$rownum<-seq(1,nrow(HD_MozB_HE),1)
HD_MozB_HE$Dstat<-HD_MozB_HE$Numer/HD_MozB_HE$Denom
```

```r
mean(HD_MozB_HE$Dstat)
```

```r
manplt<-ggplot(data = HD_MozB_HE[HD_MozB_HE$Dstat>mean(HD_MozB_HE$Dstat),], aes(x = rownum, y = Dstat, color = CHR,alpha=0.25)) +
  geom_point(size = 3) +
#  scale_color_viridis_d(option='turbo') +
theme_minimal()+
theme(axis.title = element_text(size = 18),axis.text = element_text(size = 14))
show(manplt)
max(HD_MozB_HE$Dstat)
```
